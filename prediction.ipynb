```python
import pandas as pd

# Assuming new data is in a CSV file
new_data = pd.read_csv("final.csv")

X_new = new_data[["cum_runs", "cum_wickets", "current_run_rate", "required_run_rate", "target", "batting_team_encoded", "bowling_team_encoded", "venue_canonical_encoded"]]
```

```python
import joblib
import numpy as np

# Load the saved model
model = joblib.load("final_rf_model.pkl")

# Load the Label Encoder (if used for team names)
team_encoder = joblib.load("le_team.pkl")  # If applicable
venue_encoder = joblib.load("le_venue.pkl")
```

```python
# Take user input
user_input = {
    "inning": input("Enter inning: "),
    "batting_team": input("Enter Batting Team: "),
    "bowling_team": input("Enter Bowling Team: "),
    "cum_runs": float(input("Enter Cumulative Runs: ")),
    "cum_wickets": int(input("Enter Cumulative Wickets: ")),
    "current_run_rate": float(input("Enter Current Run Rate: ")),
    "required_run_rate": float(input("Enter Required Run Rate: ")),
    "target": int(input("Enter Target Score: ")),
    "venue": input("Enter Venue: ")
}

# ---------------------------
# 3. Encode Categorical Variables
# ---------------------------
# Encode batting team and bowling team using the team encoder
batting_team_encoded = team_encoder.transform([user_input['batting_team']])[0]
bowling_team_encoded = team_encoder.transform([user_input['bowling_team']])[0]

# Encode the venue using the venue encoder
venue_canonical_encoded = venue_encoder.transform([user_input['venue']])[0]
```

```python
# Enter Match ID:  335982
# Enter Batting Team:  Kolkata Knight Riders
# Enter Bowling Team:  Royal Challengers Bangalore
# Enter Cumulative Runs:  1
# Enter Cumulative Wickets:  0
# Enter Current Run Rate:  0
# Enter Required Run Rate:  11.1
# Enter Target Score:  223
# Enter Venue:  M Chinnaswamy Stadium
```

```python
input_data = {
    'inning': [user_input['inning']],
    'cum_runs': [user_input['cum_runs']],
    'cum_wickets': [user_input['cum_wickets']],
    'current_run_rate': [user_input['current_run_rate']],
    'required_run_rate': [user_input['required_run_rate']],
    'target': [user_input['target']],
    'batting_team_encoded': [batting_team_encoded],
    'bowling_team_encoded': [bowling_team_encoded],
    'venue_canonical_encoded': [venue_canonical_encoded]
}

input_df = pd.DataFrame(input_data)
```

```python
prediction = model.predict(input_df)[0]
predicted_probabilities = model.predict_proba(input_df)[0]

# ---------------------------
# 6. Map Prediction to Team Names
# ---------------------------
if prediction == 1:
    predicted_winner = user_input['batting_team']
else:
    predicted_winner = user_input['bowling_team']

print("Predicted Winner:", predicted_winner)
print("Prediction Probabilities (Loss, Win):", predicted_probabilities)
```
