```python
import pandas as pd
```

```python
matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")
```

```python
print("Matches Dataset Info:")
matches.info()
print("\nDeliveries Dataset Info:")
deliveries.info()
```

```python
matches.rename(columns={'id': 'match_id'}, inplace=True)
```

```python
drop_columns_in_matches = ['match_type', 'player_of_match', 'target_runs',
                           'target_overs', 'super_over', 'method', 'umpire1', 'umpire2', 'venue']
matches.drop(columns=drop_columns_in_matches, inplace=True, errors='ignore')
```

```python
matches['city'].fillna('Unknown', inplace=True)
matches['winner'].fillna("no result", inplace=True)
matches['result_margin'].fillna(0, inplace=True)
deliveries.fillna("Not Available", inplace=True)
```

```python
print("\nNull values in Matches dataset:")
print(matches.isnull().sum())
print("\nNull values in Deliveries dataset:")
print(deliveries.isnull().sum())
```

```python
matches['match_id'] = matches['match_id'].astype(str)
deliveries['match_id'] = deliveries['match_id'].astype(str)
```

```python
merged_data = deliveries.merge(matches, on='match_id', how='inner')
```

```python
print("\nMerged Dataset Sample:")
print(merged_data.head())
```

```python
merged_data.to_csv("cleaned_merged_data.csv", index=False)
print("\nCleaned dataset saved as 'cleaned_merged_data.csv'")
```

```python
cleaned_data = pd.read_csv("cleaned_merged_data.csv")

print("\nNull values in Cleaned Dataset:")
print(cleaned_data.isnull().sum())
```
