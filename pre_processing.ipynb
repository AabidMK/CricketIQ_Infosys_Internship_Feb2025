```python
df1['extras_type'].fillna('no extras', inplace=True)
df1['player_dismissed'].fillna('not dismissed', inplace=True)
df1['dismissal_kind'].fillna('no dismissal', inplace=True)
df1['fielder'].fillna('no fielding', inplace=True)
```

```python
df2.head()
```

```python
df2.isnull().sum()
```

```python
df2['winner'].fillna('no result', inplace=True)
df2['player_of_match'].fillna('no one', inplace=True)
df2['result_margin'].fillna(0, inplace=True)
```

```python
df1.duplicated()
```

```python
df2.duplicated()
```

```python
selected_columns_df1 = ['match_id', 'inning', 'batting_team', 'bowling_team', 'over', 'ball', 'total_runs', 'is_wicket']
selected_columns_df2 = ['id', 'date', 'team1', 'team2', 'toss_winner', 'toss_decision', 'winner', 'result', 'result_margin']

merged_df = pd.merge(df1[selected_columns_df1], df2[selected_columns_df2], 
                     left_on='match_id', right_on='id')
print(merged_df)
```

```python
merged_df.isnull().sum()
```

```python
merged_df['Cumulative_runs']=merged_df.groupby(["match_id","inning"])['total_runs'].cumsum()
print("\Cumulative runs")
print(merged_df[['match_id','inning','Cumulative_runs']].head())
```

```python
merged_df['cumulative_wickets'] = merged_df.groupby(['match_id', 'inning'])['is_wicket'].cumsum()
print("\Cumulative Wickets")
print(merged_df[['match_id', 'inning', 'cumulative_wickets']].head())
```

```python
merged_df['cumulative_runs'] = merged_df.groupby(['match_id', 'inning'])['total_runs'].cumsum()
merged_df['overs_faced'] = merged_df['over'] + (merged_df['ball'] - 1) / 6
merged_df['overs_faced'].replace(0, 0.1, inplace=True) 
merged_df['current_run_rate'] = merged_df['cumulative_runs'] / merged_df['overs_faced']
print(merged_df[['match_id', 'inning', 'over', 'ball', 'total_runs', 'cumulative_runs', 'overs_faced', 'current_run_rate']].head())
```

```python
merged_df['cumulative_runs'] = merged_df.groupby(['match_id', 'inning'])['total_runs'].cumsum()
merged_df['overs_faced'] = merged_df['over'] + (merged_df['ball'] - 1) / 6
merged_df['overs_faced'].replace(0, 0.1, inplace=True) 
merged_df['overs_completed'] = merged_df['over'] + merged_df['ball'] / 6
print(merged_df[['match_id', 'inning', 'over', 'ball', 'total_runs', 'overs_faced', 'overs_completed']])
```

```python
target_score = merged_df[merged_df['inning'] == 1]['total_runs'].sum() + 1
remaining_runs = target_score - merged_df['cumulative_runs']
remaining_overs = 20 - merged_df['overs_completed']  # Assuming a 20-over match
merged_df['required_run_rate'] = remaining_runs / remaining_overs
print(merged_df[['match_id', 'inning', 'over', 'ball', 'total_runs', 'cumulative_runs', 'overs_faced', 'overs_completed', 'required_run_rate']])
```
