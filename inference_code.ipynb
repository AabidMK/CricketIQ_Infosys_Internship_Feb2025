```python
import joblib
import numpy as np
import pandas as pd
from google.colab import drive

# Step 1: Mount Google Drive
drive.mount('/content/drive')

# Step 2: Define the correct path for the trained model
drive_path = "/content/drive/My Drive/ipl_cricket"


# Step 3: Load the trained model
model = joblib.load(drive_path + "/final_rf_model.pkl")

# Step 4: Define team encoding (must match training data)
team_encoding = {
    'Mumbai Indians': 0,
    'Chennai Super Kings': 1,
    'Royal Challengers Bengaluru': 2,
    'Kolkata Knight Riders': 3,
    'Delhi Capitals': 4,
    'Sunrisers Hyderabad': 5,
    'Punjab Kings': 6,
    'Rajasthan Royals': 7,
    'Gujarat Titans': 8,
    'Lucknow Super Giants': 9
}

# Function to preprocess match data for inference
def preprocess_match_data(batting_team, bowling_team, inning, runs, wickets, crr, rrr, target, venue_encoded):
    """
    Converts match details into a format suitable for model inference.
    """
    # Ensure the team names exist in the encoding
    if batting_team not in team_encoding or bowling_team not in team_encoding:
        raise ValueError("Unknown team name! Check the team list used during training.")

    # Convert data to DataFrame
    match_df = pd.DataFrame([{
        'inning': inning,
        'cumulative_runs': runs,
        'cumulative_wickets': wickets,
        'current_run_rate': crr,
        'required_run_rate': rrr,
        'target_runs': target,
        'batting_team_encoded': team_encoding[batting_team],
        'bowling_team_encoded': team_encoding[bowling_team],
        'venue_encoded': venue_encoded  # Venue encoding should match training
    }])

    return match_df

# Example Match Scenario: Mumbai Indians vs RCB
match_data = preprocess_match_data(
    batting_team="Mumbai Indians",
    bowling_team="Royal Challengers Bengaluru",
    inning=2,  # MI batting second
    runs=120,  # Current score
    wickets=4,  # Wickets lost
    crr=8.5,  # Current Run Rate
    rrr=9.0,  # Required Run Rate
    target=180,  # Target score
    venue_encoded=3  # Example venue encoding
)

# Step 5: Perform inference (Predict winner)
prediction = model.predict(match_data)[0]

# Step 6: Output the result
if prediction == 1:
    print("\n🏆 Prediction: Mumbai Indians are likely to WIN!")
else:
    print("\n❌ Prediction: Mumbai Indians are likely to LOSE, so RCB might WIN!")
```
