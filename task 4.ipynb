```python
import pandas as pd
import numpy as np
df=pd.read_csv('filtered.csv')
```

```python
if 'winner' in df.columns:
    df['win'] = (df['batting_team'] == df['winner']).astype(int)

```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()


# Encode the 'batting_team' column
encoded_batting_teams = label_encoder.fit_transform(df['batting_team'])

# Add the encoded values as a new column
df['batting_team_encoding'] = encoded_batting_teams

# Display the encoding mapping
batting_team_mapping = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
print('Batting Team Encoding Mapping:', batting_team_mapping)
```

```python
df
```

```python
encoded_bowling_teams = label_encoder.fit_transform(df['bowling_team'])
df['bowling_team_encoding'] = encoded_bowling_teams
```

```python
df
```

```python
df.drop(['batting_team', 'bowling_team'], axis=1, inplace=True)
df['Required Run Rate'] = pd.to_numeric(df['Required Run Rate'], errors='coerce').fillna(0)
```

```python
categorical_columns = df.select_dtypes(include=['object']).columns
for column in categorical_columns:
    df[column] = label_encoder.fit_transform(df[column])
```

```python

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df['winner'])

X = df.drop(['winner'], axis=1)

# Split data into training and testing sets (70-30)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize and train a Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f'Model Accuracy: {accuracy * 100:.2f}%')
```
