importing libraries 

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder #for encoding
from sklearn.model_selection import train_test_split #for training and testing
```

reading merged dataset

```python
merged=pd.read_csv(r"merged_dataset.csv",low_memory=False)
merged
```

renaming the coulumns cumulative runs and cumulatives wickets

```python
merged.rename(columns={
    'cumulative runs': 'cum_runs',
    'cumumulative wickets': 'cum_wickets'
}, inplace=True)
```

columns after renaming cumulative runs and cumulatives wickets

```python
merged.columns
```

using labelencoder we are encoding the categorical values like batting team,bowling team,venue

```python
label_encoder = LabelEncoder()
merged['batting_team_encoded'] = label_encoder.fit_transform(merged['batting_team'])
merged['bowling_team_encoded'] = label_encoder.fit_transform(merged['bowling_team'])
merged['venue_encoded'] = label_encoder.fit_transform(merged['venue'])
```

binary feature for win 

```python
merged['win'] = (merged['batting_team'] == merged['winner']).astype(int)
```

merged dataset after encoding categorical values and binary feature for win 

```python
merged.head()
```

only these columns for final dataframe

```python
final_df = merged[['inning', 'cum_runs', 'cum_wickets', 'batting_team_encoded', 
               'bowling_team_encoded', 'venue_encoded', 'win']]
final_df.to_csv('final_dataframe.csv', index=False)
```

```python
final_df.columns
```

```python
X = final_df.drop('win', axis=1)
Y = final_df['win']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("traning and testing done successfully")
```
