```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Merged Data.csv')
```

```python
df.head() #first above five rows from the Df
```

```python
df = df.drop(columns=['remaining_overs','overs_completed'])
```

```python
#Checking Null Value in The Dataframe
Null_Counts = df.isnull().sum()
print(Null_Counts)
```

```python
#fill the null values in the required run rate column
df = df.fillna({'required_run_rate':'0'})
print(df)
```

```python
print(df["required_run_rate"].dtype)
```

```python

df["required_run_rate"] = df["required_run_rate"].astype(float)
print(df["required_run_rate"].dtype)
```

```python
#check null after fill na function
Null = df.isnull().sum()
print(Null)
```

```python
df['win'] = (df['winner'] == df['batting_team']).astype(int)
df = df.drop(columns=['winner'])
```

```python
print("New class distribution:\n", df['win'].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
team_encoder = LabelEncoder()
venue_encoder = LabelEncoder()
```

```python
# Fit the encoder on all unique teams and apply it to both batting and bowling teams
df['batting_team_encoded'] = team_encoder.fit_transform(df['batting_team'])
df['bowling_team_encoded'] = df['bowling_team'].map(lambda x: team_encoder.transform([x])[0])
```

```python
df['venue_encoded'] = venue_encoder.fit_transform(df['venue'])
```

```python
# Check if batting and bowling teams have the same encoding values
print(df[['batting_team', 'batting_team_encoded', 'bowling_team', 'bowling_team_encoded']])
```

```python
# Drop original categorical columns
df = df.drop(columns=['batting_team', 'bowling_team', 'venue'])
```

```python
categorical_columns_data = df.select_dtypes(include=['object', 'category']).columns
print("Categorical Columns Data:", categorical_columns_data.tolist())
```

```python
import pickle
# Save encoders for future use
with open('/content/drive/My Drive/Colab Notebooks/team_encoder.pkl', "wb") as f:
    pickle.dump(team_encoder, f)

with open('/content/drive/My Drive/Colab Notebooks/venue_encoder.pkl', "wb") as f:
    pickle.dump(venue_encoder, f)

print("Preprocessing completed successfully!")
```

```python
#Train Test Split on Df
X= df.drop(columns=['win'])#independent var
Y = df['win'] #dependent var
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size=0.2, random_state=42)
```

```python
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
#For check the target column or Predictor
print(X_train.columns,'\n')
print(X_test.columns,"\n")
print(Y_train.name,'\n')
print(Y_test.name,'\n')
```

```python
#Train & Test Features Of X-Axis
print("X Train Features",X_train,"\n")
print("X Test Fetures",X_test,"\n")
```

```python
#Train & Test Labels Of Y-Axis
print("Y Train Labels:\n",Y_train,"\n")
print("Y Test Labels:\n",Y_test,"\n")
```

```python
#Random Forest
from sklearn.ensemble import RandomForestClassifier
```

```python
# Initialize Random Forest model
random_forest_model = RandomForestClassifier(n_estimators = 200 , max_depth = 15, min_samples_split = 10  ,min_samples_leaf = 4, max_features = 'sqrt', bootstrap = True , random_state=42)
#After using Hyperparameters Model is Balanced not Overfitted
# Fit the model on training data
#random_forest_model.fit(X_train, Y_train)
```

```python
#Classification Report, Accuracy, Confusion Matrix -- Metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
from imblearn.over_sampling import RandomOverSampler

random_over_sample = RandomOverSampler(random_state=42)

X_train_resampled, Y_train_resampled = random_over_sample.fit_resample(X_train, Y_train)
```

```python
model = RandomForestClassifier(max_depth=15, n_estimators=200)
model.fit(X_train_resampled, Y_train_resampled)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
param_distributions = {
   'n_estimators': [100, 200, 300],
   'max_depth': [5, 10, 15],
   'min_samples_split': [2, 5, 10],
   'min_samples_leaf': [1, 2, 4]
}
rf_random = RandomizedSearchCV(estimator=RandomForestClassifier(),
                             param_distributions=param_distributions,
                             cv=3, n_iter=10, random_state=42)
rf_random.fit(X_train_resampled, Y_train_resampled)
best_params = rf_random.best_params_
```

Model Performance

```python
# Predict on training and testing sets
random_forest_train_prediction = random_forest_model.predict(X_train_resampled)
random_forest_test_prediction = random_forest_model.predict(X_test)
```

```python
Train_data = accuracy_score(Y_train_resampled, random_forest_train_prediction)

Testing_data = accuracy_score(Y_test, random_forest_test_prediction)
```

```python
random_forest_conf_matrix = confusion_matrix(Y_test, random_forest_test_prediction)

labels = ["Win (1)", "Loss (0)"]
```

```python
print(f"Train Accuracy : {Train_data:.2f}")
print(f"Test Accuracy : {Testing_data:.2f}")

#classification Report
print("\nClassification Report Testing:\n", classification_report(Y_test, random_forest_test_prediction))
#confusion matrix
print("\nConfusion Matrix Testing:\n", random_forest_conf_matrix)
```
