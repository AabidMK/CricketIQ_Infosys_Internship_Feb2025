```python
#Drop columns
Data1_df.drop(columns=['match_type', 'player_of_match', 'target_runs', 'target_overs',
                'super_over', 'umpire1', 'umpire2', 'season', 'city', 'date',
                'toss_winner', 'toss_decision', 'result_margin', 'result', 'method',
                'team1', 'team2'], inplace=True)
```

```python
#Drop columns
Data2_df.drop(columns=['batter', 'bowler', 'non_striker', 'batsman_runs', 'extra_runs', 'extras_type', 'player_dismissed','fielder','dismissal_kind'], inplace=True)
Data2_df.head()
```

```python
#check Null Data
Data2_null = Data2_df.isnull().sum()
print(Data2_null)
```

```python
team_name_mapping = {
    'Kings XI Punjab': 'Punjab Kings',
    'Delhi Daredevils': 'Delhi Capitals',
    'Chennai Super Kings': 'CSK',
    'Royal Challengers Bangalore' : 'RCB',
    'Rising Pune Supergiants' : 'Rising Pune Supergiants',
    'Kolkata Knight Riders' : 'Knight Riders',
    'Rajasthan Royals' : 'RR',
}


Data1_df['winner']=Data1_df['winner'].replace(team_name_mapping)
Data1_df.head()
```

```python
print("Data 2")
Data2_df.head()
```

```python
# Merged Both Data1 and Data2
merged_data = pd.merge(Data2_df, Data1_df, left_on='match_id', right_on='id', how='left')
print("Merged Data (Deliveries Data & Matches Data)\n")
merged_data = merged_data.drop(columns=['id'])
merged_data.head()
```

```python
merged_data.info()
```

```python
#Nul Counts Merged Data
null = merged_data.isnull().sum()
print(null)
```

```python
# Cumulative Runs
merged_data["cumulative_runs"] = merged_data.groupby(['match_id','inning'])['total_runs'].cumsum()
print(merged_data)

```

```python
# Overs Completed 

merged_data["overs_completed"] =  merged_data['over'] + (merged_data['ball'] - 1) / 6

print(merged_data)
```

```python
#Cumulative Wickets

merged_data["cumulative_wickets"] = merged_data.groupby(['match_id','inning'])['is_wicket'].cumsum()
print(merged_data)

```

```python
#Current RuN Rate

import numpy as np

merged_data['current_run_rate'] = np.where(merged_data['overs_completed']== 0,0,merged_data['cumulative_runs']/merged_data['overs_completed'])

print(merged_data)
```

```python
merged_data.head()
```

```python
# Target and Required Run Rate 
#First Inning
first_inning = merged_data[merged_data['inning'] == 1]
cum_runs = first_inning['cumulative_runs'].max()
target = cum_runs + 1
print("Target:", target)

total_over = first_inning['over'].max()

print("Total Over In First Inning :",total_over)



required_run_rate = target/total_over
print("Required Run Rate",required_run_rate )
merged_data['target'] = target
```

```python
 #Remaining Overs

total_overs = 20

merged_data['remaining_overs'] = total_overs - merged_data['overs_completed']

print(merged_data)
```

```python
# Second Inning
second_inning = merged_data[merged_data['inning'] == 2].copy()
second_inning['required_run_rate'] = (merged_data['target'] - second_inning['cumulative_runs']) / second_inning['remaining_overs']

# Replace infinite values with 0
second_inning['required_run_rate'] = second_inning['required_run_rate'].replace([np.inf, -np.inf], 0)
second_inning['required_run_rate'] = np.where(second_inning['remaining_overs'] <= 0, 0, second_inning['required_run_rate'])

merged_data['required_run_rate'] = second_inning['required_run_rate'] 

print(merged_data.head)
```

```python
merged_data.head()
```

```python
#Drop Unnecessary columns

Merged_Data = merged_data.drop(columns=['match_id','over','ball','is_wicket','total_runs'])

print(Merged_Data.head())

```

```python
Merged_Data['win'] = np.where(Merged_Data['batting_team'] == Merged_Data['winner'],1,0)
second_inning = Merged_Data[Merged_Data['inning']==2]
print(second_inning.head())
```

```python
#Drop Winner Column 
Merged_Data = Merged_Data.drop(columns=['winner'])
print(Merged_Data)
```
