matches dataset

dropping unnecessary columns in matches

```python
drop_columns_in_matches=['match_type','player_of_match','target_runs','target_overs','super_over','method','umpire1','umpire2']
matches.drop(columns=drop_columns_in_matches,inplace=True,errors='ignore')
```

```python
print(matches.columns)
```

renaming team columns 

```python
rename_team = {
    'team1':'home_team',
    'team2':'away_team'
}
matches.rename(columns=rename_team,inplace=True)
```

after renaming teams 

```python
print(matches.columns)
```

mapping venue to cities for missing values 

```python
venue_city = {
    'Dubai International Cricket Stadium': 'Dubai',
    'Sharjah Cricket Stadium': 'Sharjah',
    'Sheikh Zayed Stadium': 'Abu Dhabi'
}

matches['city'] = matches.apply(lambda x: venue_city.get(x['venue'], x['city']), axis=1)
matches['city'].fillna('Unknown', inplace=True)
```

filling missing values of winner and result margin

```python
matches['winner'].fillna("no reasult",inplace=True)
```

```python
matches['result_margin'].fillna(0,inplace=True)
```

standarlizing team name in mathces 

```python
team_name = {
    'Kings XI Punjab': 'Punjab Kings',
    'Delhi Daredevils': 'Delhi Capitals',
    'Deccan Chargers': 'Sunrisers Hyderabad'
}

matches['home_team'] = matches['home_team'].replace(team_name)
matches['away_teeam'] = matches['away_team'].replace(team_name)
matches['winner'] = matches['winner'].replace(team_name)
matches['toss_winner'] = matches['toss_winner'].replace(team_name)
```

deliveries dataset 

only the columns whice are useful are taken from deliveries dataset 

```python
deliveries = deliveries[['match_id', 'inning', 'batting_team', 'bowling_team', 'over', 'ball', 'total_runs', 'is_wicket']]
```

standardizing team name in deliveries 

```python
deliveries = deliveries.copy()
deliveries['batting_team'] = deliveries['batting_team'].replace(team_name)
deliveries['bowling_team'] = deliveries['bowling_team'].replace(team_name)
```

calculating cumilative runs,wickets

```python
deliveries['cumulative runs'] = deliveries.groupby(['match_id', 'inning'])['total_runs'].cumsum()
deliveries['cumumulative wickets'] = deliveries.groupby(['match_id', 'inning'])['is_wicket'].cumsum()
```

overs completed

```python
deliveries['over_completed']= deliveries['over']+(deliveries['ball']-1)/6
```

current run rate 

```python
deliveries['current run rate'] = deliveries['cumulative runs']/deliveries['over_completed']
deliveries['current run rate'].fillna(0,inplace=True)
```

target and run rate for second innings

```python
first_innings_scores = deliveries[deliveries['inning'] == 1].groupby('match_id')['cumulative runs'].max().reset_index()
first_innings_scores.rename(columns={'cumulative runs': 'target_score'}, inplace=True)
```

```python
first_innings_scores['target'] = first_innings_scores['target_score'] + 1
```

merging target in to deliveries

```python
deliveries = deliveries.merge(first_innings_scores[['match_id', 'target']], on='match_id', how='left')
```

run rate for second innings

```python
deliveries['remaining_overs'] = 20 - deliveries['over_completed']
deliveries['required_run_rate'] = (deliveries['target'] - deliveries['cumulative runs']) / deliveries['remaining_overs']
deliveries['required_run_rate'].fillna(0, inplace=True)
```

merge matches and deliveries data 

```python
merged = deliveries.merge(matches, left_on='match_id', right_on='id', how='left')
merged.drop(columns=['id'], inplace=True)  
```
