importing libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
final_df=pd.read_csv(r"final_dataframe.csv")
final_df
```

training and testing the dataframe

```python
X = final_df.drop('win', axis=1)
Y = final_df['win']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("traning and testing done successfully")
```

evaluating the models

importing libraries for models 

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

logistic regression 

```python
print("training and testing for logistic regression")
logistic_regression = LogisticRegression()
logistic_regression.fit(X_train,Y_train)
Y_train_pred=logistic_regression.predict(X_train)
Y_test_pred=logistic_regression.predict(X_test)
print("done successfully")
print("=======================================================")
print("performance on training data\n")
print("=======================================================")
print("accuracy:", accuracy_score(Y_train, Y_train_pred))
print("-------------------------------------------------------")
print("classification report:\n", classification_report(Y_train, Y_train_pred))
print("-------------------------------------------------------")
print("confusion matrix:\n", confusion_matrix(Y_train, Y_train_pred))
print("=======================================================")
print("performance on test data\n")
print("=======================================================")
print("accuracy:", accuracy_score(Y_test, Y_test_pred))
print("-------------------------------------------------------")
print("classification report:\n", classification_report(Y_test, Y_test_pred))
print("-------------------------------------------------------")
print("confusion matrix:\n", confusion_matrix(Y_test, Y_test_pred))
```

random forest

```python
print("training and testing for random forest")
random_forest = RandomForestClassifier()
random_forest.fit(X_train,Y_train)
Y_train_pred=random_forest.predict(X_train)
Y_test_pred=random_forest.predict(X_test)
print("done successfully")
print("=======================================================")
print("performance on training data\n")
print("=======================================================")
print("accuracy:", accuracy_score(Y_train, Y_train_pred))
print("-------------------------------------------------------")
print("classification report:\n", classification_report(Y_train, Y_train_pred))
print("-------------------------------------------------------")
print("confusion matrix:\n", confusion_matrix(Y_train, Y_train_pred))
print("=======================================================")
print("performance on test data\n")
print("=======================================================")
print("accuracy:", accuracy_score(Y_test, Y_test_pred))
print("-------------------------------------------------------")
print("classification report:\n", classification_report(Y_test, Y_test_pred))
print("-------------------------------------------------------")
print("confusion matrix:\n", confusion_matrix(Y_test, Y_test_pred))
```

```python
print("training and testing for XGBoost")
xgboost = XGBClassifier()
xgboost.fit(X_train,Y_train)
Y_train_pred=xgboost.predict(X_train)
Y_test_pred=xgboost.predict(X_test)
print("done successfully")
print("=======================================================")
print("performance on training data\n")
print("=======================================================")
print("accuracy:", accuracy_score(Y_train, Y_train_pred))
print("-------------------------------------------------------")
print("classification report:\n", classification_report(Y_train, Y_train_pred))
print("-------------------------------------------------------")
print("confusion matrix:\n", confusion_matrix(Y_train, Y_train_pred))
print("=======================================================")
print("performance on test data\n")
print("=======================================================")
print("accuracy:", accuracy_score(Y_test, Y_test_pred))
print("-------------------------------------------------------")
print("classification report:\n", classification_report(Y_test, Y_test_pred))
print("-------------------------------------------------------")
print("confusion matrix:\n", confusion_matrix(Y_test, Y_test_pred))
```
