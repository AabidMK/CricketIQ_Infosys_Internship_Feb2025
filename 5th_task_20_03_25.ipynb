```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
final_df=pd.read_csv(r"/content/Data set final.csv")
final_df
```

```python
X = final_df.drop('winner', axis=1)
Y = final_df['winner']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
print("Training and testing done successfully")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Separate numeric and categorical columns
numeric_cols = X_train.select_dtypes(include=['number']).columns
categorical_cols = X_train.select_dtypes(include=['object']).columns

# Ensure all numeric columns are properly converted to float
X_train[numeric_cols] = X_train[numeric_cols].apply(pd.to_numeric, errors='coerce')
X_test[numeric_cols] = X_test[numeric_cols].apply(pd.to_numeric, errors='coerce')

# Fill NaNs in numeric columns with median
X_train[numeric_cols] = X_train[numeric_cols].fillna(X_train[numeric_cols].median())
X_test[numeric_cols] = X_test[numeric_cols].fillna(X_test[numeric_cols].median())

# Fill NaNs in categorical columns with 'Unknown'
X_train[categorical_cols] = X_train[categorical_cols].fillna('Unknown')
X_test[categorical_cols] = X_test[categorical_cols].fillna('Unknown')

# Standardize the numeric columns
scaler = StandardScaler()
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])

# Encode categorical columns
label_encoders = {}
for col in categorical_cols:
    label_encoders[col] = LabelEncoder()

    # Fit encoder on combined data to avoid unseen labels during testing
    combined_data = pd.concat([X_train[col], X_test[col]], axis=0)
    label_encoders[col].fit(combined_data)

    X_train[col] = label_encoders[col].transform(X_train[col])
    X_test[col] = label_encoders[col].transform(X_test[col])

# Train Logistic Regression model
print("\nTraining Logistic Regression model with proper NaN handling, encoding, and scaling...")
logistic_regression = LogisticRegression(max_iter=1500, solver='lbfgs', random_state=42)
logistic_regression.fit(X_train, Y_train)
print("Model trained successfully!")

# Predictions
Y_train_pred = logistic_regression.predict(X_train)
Y_test_pred = logistic_regression.predict(X_test)

# Display performance metrics
def print_performance(name, y_true, y_pred):
    print(f"\n{name} Performance Metrics")
    print("Accuracy:", accuracy_score(y_true, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

# Display results
print_performance("Training", Y_train, Y_train_pred)
print_performance("Test", Y_test, Y_test_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder

# Fill missing values with cumulative fill
X_train = X_train.ffill().bfill()
X_test = X_test.ffill().bfill()

# Encode categorical columns
categorical_cols = X_train.select_dtypes(include='object').columns

# Apply LabelEncoder to all categorical columns
label_encoders = {}
for col in categorical_cols:
    label_encoders[col] = LabelEncoder()

    # Fit on combined data to avoid unseen labels during testing
    combined_data = pd.concat([X_train[col], X_test[col]], axis=0)

    label_encoders[col].fit(combined_data)

    # Transform both training and testing sets
    X_train[col] = label_encoders[col].transform(X_train[col])
    X_test[col] = label_encoders[col].transform(X_test[col])

# Fit Random Forest model
print("\nTraining and testing for Random Forest...")
random_forest = RandomForestClassifier(n_estimators=100, random_state=42)
random_forest.fit(X_train, Y_train)
print("Model trained successfully!")

# Predictions
Y_train_pred = random_forest.predict(X_train)
Y_test_pred = random_forest.predict(X_test)

# Display performance metrics
def print_performance(name, y_true, y_pred):
    print(f"\n{name} Performance Metrics")
    print("Accuracy:", accuracy_score(y_true, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

# Display results
print_performance("Training", Y_train, Y_train_pred)
print_performance("Test", Y_test, Y_test_pred)
```

```python
from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Encode target labels
print("\nEncoding target labels...")
label_encoder = LabelEncoder()

# Fit on combined target labels to avoid unseen classes during testing
combined_y = pd.concat([Y_train, Y_test], axis=0)
label_encoder.fit(combined_y)

# Transform labels
Y_train_encoded = label_encoder.transform(Y_train)
Y_test_encoded = label_encoder.transform(Y_test)

# Train XGBoost model
print("\nTraining and testing for XGBoost...")
xgboost = XGBClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
xgboost.fit(X_train, Y_train_encoded)
print("Model trained successfully!")

# Predictions
Y_train_pred = xgboost.predict(X_train)
Y_test_pred = xgboost.predict(X_test)

# Decode predictions back to original labels for better readability
Y_train_pred_decoded = label_encoder.inverse_transform(Y_train_pred)
Y_test_pred_decoded = label_encoder.inverse_transform(Y_test_pred)

# Display performance metrics
def print_performance(name, y_true, y_pred):
    print(f"\n{name} Performance Metrics")
    print("Accuracy:", accuracy_score(y_true, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

# Display results
print_performance("Training", Y_train, Y_train_pred_decoded)
print_performance("Test", Y_test, Y_test_pred_decoded)
```
