```python
for col in matches.columns:
    if matches[col].dtype == 'object':
        matches[col].fillna(matches[col].mode()[0], inplace=True)
    else:
        matches[col].fillna(matches[col].median(), inplace=True)

print("\nMissing Values After Handling:\n", matches.isnull().sum())
```

```python
for col in deliveries.columns:
    if deliveries[col].dtype == 'object':
        deliveries[col].fillna(deliveries[col].mode()[0], inplace=True)
    else:
        deliveries[col].fillna(deliveries[col].median(), inplace=True)

print("\nMissing Values After Handling:\n", deliveries.isnull().sum())
```

```python
print("Duplicate rows:", matches.duplicated().sum())
print("Duplicate rows:",deliveries.duplicated().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

# Label encode categorical columns
label_cols = ['team1', 'team2', 'toss_winner', 'winner', 'venue']

for col in label_cols:
    le = LabelEncoder()
    matches[col] = le.fit_transform(matches[col])
```

```python
# Drop unnecessary columns
columns_to_drop = ['match_id', 'series_id', 'date', 'match_description', 'umpire1', 'umpire2']
df = matches.drop(columns=columns_to_drop, errors='ignore')
```

```python
print(matches.columns)
print(matches.head())
print(deliveries.head())
print(deliveries.columns)
```

```python
matches_filtered = matches[["id", "date", "team1", "team2", "toss_winner", "toss_decision", "winner", "result", "result_margin"]]
```

```python
matches_filtered = matches.drop(columns=["match_type", "player_of_match", "target_runs"], errors='ignore')
```

```python
matches_filtered.rename(columns={"id": "match_id", "team1": "home_team", "team2": "away_team"}, inplace=True)
```

```python
venue_city_mapping = {"Dubai International Cricket Stadium": "Dubai", "Sharjah Cricket Stadium": "Sharjah"}
matches_filtered["city"] = matches_filtered["venue"].map(venue_city_mapping).fillna(matches_filtered["city"])
```

```python
matches_filtered["winner"] = matches_filtered["winner"].fillna("no result")
matches_filtered["result_margin"] = matches_filtered["result_margin"].fillna(0)

```

```python
team_mapping = {"Kings XI Punjab": "Punjab Kings", "Delhi Daredevils": "Delhi Capitals"}
matches_filtered.replace({"home_team": team_mapping, "away_team": team_mapping, "winner": team_mapping}, inplace=True)
```

```python

deliveries_filtered = deliveries[["match_id", "inning", "batting_team", "bowling_team", "over", "ball", "total_runs", "is_wicket"]].copy()


deliveries_filtered = deliveries_filtered.replace({"batting_team": team_mapping, "bowling_team": team_mapping})
```

```python
deliveries_filtered = deliveries_filtered.copy()

deliveries_filtered.loc[:, "cumulative_runs"] = deliveries_filtered.groupby(["match_id", "inning"])["total_runs"].cumsum()
deliveries_filtered.loc[:, "cumulative_wickets"] = deliveries_filtered.groupby(["match_id", "inning"])["is_wicket"].cumsum()
```

```python
deliveries_filtered.loc[:, "overs_completed"] = deliveries_filtered["over"] + (deliveries_filtered["ball"] - 1) / 6
deliveries_filtered.loc[:, "current_runrate"] = deliveries_filtered["cumulative_runs"] / deliveries_filtered["overs_completed"]
deliveries_filtered.loc[:, "current_runrate"] = deliveries_filtered["current_runrate"].fillna(0)
```

```python
target_scores = deliveries_filtered[deliveries_filtered["inning"] == 1].groupby("match_id")["cumulative_runs"].max().reset_index()
target_scores.rename(columns={"cumulative_runs": "target_score"}, inplace=True)
target_scores["target_score"] += 1
```

```python
deliveries_filtered = deliveries_filtered.merge(target_scores, on="match_id", how="left")
```

```python
deliveries_filtered["remaining_overs"] = 20 - deliveries_filtered["overs_completed"]
```

```python

deliveries_filtered["remaining_overs"] = 20 - deliveries_filtered["overs_completed"]
deliveries_filtered["remaining_overs"] = deliveries_filtered["remaining_overs"].fillna(0)  # Ensure no NaN values


second_innings_mask = deliveries_filtered["inning"] == 2
deliveries_filtered.loc[second_innings_mask, "required_run_rate"] = (
    (deliveries_filtered["target_score"] - deliveries_filtered["cumulative_runs"]) / deliveries_filtered["remaining_overs"]
)


deliveries_filtered["required_run_rate"] = deliveries_filtered["required_run_rate"].replace(
    [float("inf"), -float("inf")], 0
)
deliveries_filtered["required_run_rate"] = deliveries_filtered["required_run_rate"].fillna(0)

```

```python
merged_df = deliveries_filtered.merge(matches_filtered, on="match_id", how="left")
```
