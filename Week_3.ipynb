```python
import pandas as pd
import numpy as np
```

```python
file_path = "matches.csv"
df = pd.read_csv(file_path)
```

```python
df.describe()
```

```python
df = df.drop(columns=['method','player_of_match','match_type','date','venue','umpire1','umpire2'])
```

```python
df.head()
```

```python
# Check for null values in each column
null_values = df.isnull().sum()

print("Columns with Null Values:")
print(null_values[null_values > 0])

# Display total missing values
total_missing = df.isnull().sum().sum()
print(f"\nTotal Missing Values in Dataset: {total_missing}")
```

```python
# Handling missing values

if "winner" in df.columns:
    df["winner"].fillna("Unknown", inplace=True)

if "result_margin" in df.columns:
    df["result_margin"].fillna(0, inplace=True)

if "target_runs" in df.columns:
    df["target_runs"].fillna(0, inplace=True)

if "target_overs" in df.columns:
    df["target_overs"].fillna(0, inplace=True)

#Fill any other numerical columns with median values
num_cols = df.select_dtypes(include=["number"]).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

#Check if all missing values are handled
print("Missing values after handling:\n", df.isnull().sum())

#Save the cleaned dataset (optional)
df.to_csv("deliveries_cleaned.csv", index=False)

print("\nMissing values successfully replaced with meaningful values!")
```

```python
matches = pd.read_csv('matches.csv')
deliveries = pd.read_csv('deliveries.csv')

matches.fillna({'winner': 'No Result', 'result_margin': 0, 'toss_winner': 'Unknown'}, inplace=True)
deliveries.fillna(0, inplace=True)

merged_data = pd.merge(deliveries, matches, left_on='match_id', right_on='id', how='inner')

columns_to_keep = ['match_id','date','city', 'team1', 'team2', 'toss_winner', 'toss_decision', 'winner', 'result',
                   'result_margin', 'inning', 'batting_team', 'bowling_team', 'over', 'ball',
                   'total_runs', 'is_wicket']
merged_data = merged_data[columns_to_keep]

merged_data.rename(columns={'team1': 'home', 'team2': 'away'}, inplace=True)

# Standardize team names
team_name_mapping = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Royal Challengers Bangalore': 'Royal Challengers Bengaluru',
    'Rising Pune Supergiant': 'Rising Pune Supergiants',
    "Kings XI Punjab": "Punjab Kings"
}
merged_data['home'] = merged_data['home'].replace(team_name_mapping)
merged_data['away'] = merged_data['away'].replace(team_name_mapping)
merged_data['batting_team'] = merged_data['batting_team'].replace(team_name_mapping)
merged_data['bowling_team'] = merged_data['bowling_team'].replace(team_name_mapping)
merged_data['toss_winner'] = merged_data['toss_winner'].replace(team_name_mapping)
merged_data['winner'] = merged_data['winner'].replace(team_name_mapping)

#new features
merged_data['overs_completed'] = merged_data['over'] + merged_data['ball'] / 6
merged_data['cumulative_runs'] = merged_data.groupby(['match_id', 'inning'])['total_runs'].cumsum()
merged_data['cumulative_wickets'] = merged_data.groupby(['match_id', 'inning'])['is_wicket'].cumsum()
merged_data['current_run_rate'] = merged_data['cumulative_runs'] / merged_data['overs_completed']

#required run rate
target_score = merged_data.groupby('match_id')['cumulative_runs'].transform('max')
remaining_runs = target_score - merged_data['cumulative_runs']
remaining_overs = 20 - merged_data['overs_completed']
merged_data['required_run_rate'] = remaining_runs / remaining_overs
```

```python
from sklearn.preprocessing import LabelEncoder
# Convert 'date' to datetime
merged_data['date'] = pd.to_datetime(merged_data['date'])

# Encode categorical columns using Label Encoding
label_encoder = LabelEncoder()
categorical_columns = merged_data.select_dtypes(include=['string']).columns
for column in categorical_columns:
    merged_data[column] = label_encoder.fit_transform(merged_data[column])

# Display the processed data types
print(merged_data.dtypes)
```
